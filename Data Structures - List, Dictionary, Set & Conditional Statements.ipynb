## Sets 

```python
# The output contains only {'a', 'e', 'i', 'o', 'u'} because a set automatically removes duplicate values and stores only 
# unique elements. The order of elements may vary since sets are unordered collections in Python.

my_set = {'a','e','i','o','u','a','a','i'}
my_set   
```

```python
# since the elements in set are unordered and unindexed, we cannot change the value.

my_set[4] = 's'
```

```python
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}

print(set1)
print(set2)
```

```python
set3 = set1.union(set2)
print(set3)
```

```python
set4 = set1.intersection(set2)
print(set4)
```

## Operators & Conditional Statements

Performance Category Program

```python
score = int(input("Enter a score (0-10):"))

if 0<= score <=10:
    print(score)
else:
    print("Score must be between 0 and 10")
```

```python
score = int(input("Enter a score (0-10):"))

if 0<= score <=10:
    if score> 7:
        print("Above Average: You are doing great! Keep up the efforts!!")
    elif 4<= score <= 7:
        print("Average: Good effort! Keep practicing, there's room for improvement.")
    else:
        print("Below Average: Need to Improve your performance, consistent practice will lead to better results.")
else:
    print("Score must be between 0 and 10")
```
